```python
import json

# Go up one directory ('..') to find the file in the 'backend' folder
file_path = os.path.join('..', 'initial_solution.json')

# This debug line will print the full, absolute path Python is trying to open
print(f"Attempting to open: {os.path.abspath(file_path)}")

# Open and read the JSON file
with open(file_path, 'r') as f:
    data = json.load(f)

print("\nSuccessfully loaded JSON data:")
print(data)
```

```python
# Assume 'data' is your full JSON data from the file

# --- 1. Retrieve and reformat the wave data ---
wave_data = {}

# Find all keys that start with 'wave_'
for wave_key, wave_value in data.items():
    if wave_key.startswith('wave_'):
        # Use a list comprehension to rebuild the list of drones
        # CHANGE: Only keep the 'route' key for each drone
        drones = [
            {'route': drone['route']}
            for drone in wave_value.get('drones', [])
        ]

        # Do the same for trucks
        # CHANGE: Only keep the 'route' key for each truck
        trucks = [
            {'route': truck['route']}
            for truck in wave_value.get('trucks', [])
        ]

        # Assemble the new, simplified dictionary for the current wave
        wave_data[wave_key] = {
            'drones': drones,
            'trucks': trucks
        }

# --- 2. Display the results ---
import json

# Print the newly formatted wave_data
print("--- Reformatted Wave Data ---")
print(json.dumps(wave_data, indent=2))

# Retrieve and print the summary data
summary_info = data['summary']
num_waves = summary_info['total_waves']
total_distance = summary_info['total_distance']
total_cost = summary_info['total_cost']

print("\n--- Key Summary Metrics ---")
print(f"Number of Waves: {num_waves}")
print(f"Total Distance: {total_distance}")
print(f"Total Cost: {total_cost}")
```
